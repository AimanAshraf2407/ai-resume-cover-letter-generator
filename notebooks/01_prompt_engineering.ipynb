```python
import sys
import json
import time
from pathlib import Path
from IPython.display import display, Markdown

# Add project root directory to sys.path
root_path = Path.cwd().parent
if str(root_path) not in sys.path:
    sys.path.append(str(root_path))

from models.generator import generate_resume
```

```python
profile_path = root_path / "data" / "sample_profiles" / "profile_01_software_engineer.json"
job_path = root_path / "data" / "sample_job_descriptions" / "job_01_junior_ai_engineer.txt"

with open(profile_path, "r", encoding="utf-8") as f:
    raw_profile = json.load(f)
    user_profile = json.dumps(raw_profile, indent=2)

with open(job_path, "r", encoding="utf-8") as f:
    job_description = f.read()

print("✅ Profile and Job Description loaded successfully!")
```

```python
temperatures = [0.2, 0.7]
experiment_results = {}

for temp in temperatures:
    print(f"Generating resume with temperature = {temp}...")
    start_time = time.time()
    
    output = generate_resume(
        user_profile=user_profile,
        job_description=job_description,
        temperature=temp
    )
    
    elapsed = time.time() - start_time
    experiment_results[temp] = {
        "text": output,
        "latency_seconds": round(elapsed, 2),
        "char_count": len(output),
        "word_count": len(output.split())
    }

print("\n✅ Generation complete for all temperatures!")
```

```python
for temp, data in experiment_results.items():
    display(Markdown(f"## 🌡️ Results for Temperature = {temp}"))
    display(Markdown(f"**Latency:** {data['latency_seconds']}s | **Words:** {data['word_count']}"))
    display(Markdown("---"))
    display(Markdown(data["text"]))
    display(Markdown("\n<br>\n"))
```

```python
known_skills = set(skill.lower() for skill in raw_profile.get("technical_skills", []))

print("=== FACTUALITY & SKILL ADHERENCE ANALYSIS ===")
for temp, data in experiment_results.items():
    output_lower = data["text"].lower()
    matched_skills = [skill for skill in known_skills if skill in output_lower]
    retention_rate = (len(matched_skills) / len(known_skills)) * 100
    
    print(f"\n[Temperature {temp}]")
    print(f"  • Ground-truth skills retained: {len(matched_skills)}/{len(known_skills)} ({retention_rate:.1f}%)")
    print(f"  • Retained skills: {', '.join(matched_skills)}")
```
